#### Add Memory to Chatbot

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model='openai/gpt-oss-120b'
)
```

```python
chat_history = []
```

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langchain_core.output_parsers import StrOutputParser
```

```python
out = StrOutputParser()
```

```python
chain = llm | out
```

```python
while True:
    query = input("Ask: ")

    if query in ['quit', 'byt']:
        print('Thank You')
        break

    chat_history.append(
        HumanMessage(
            content=query)
    )

    response = chain.invoke(chat_history)

    chat_history.append(
        AIMessage(content=response)
    )

    print(f"AI: {response}")
```

```python
# structured output means instructing the llm to return its response in predefined format
```

```python
from pydantic import BaseModel, Field
```

```python
class Person(BaseModel):
    name: str = Field(description='Name of the Person')
    age: int = Field(description='Age of the Person')
    city: str = Field(description='City of the Person')
```

```python
structured_llm = llm.with_structured_output(Person)
```

```python
response = structured_llm.invoke(f'Please give me name, age and city of {text}')
```

```python
print(response)
```

```python
type(response)
```

```python
response.model_dump()
```

#### AI Agent tools

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(
    model='openai/gpt-oss-120b'
)
```

```python
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
@tool
def add_numbers(a:int, b:int):
    """
    it will return sum of two numbers
    a: number one
    b: number two

    """
    return a + b
```

```python
@tool
def mul_numbers(a:int, b:int):
    """
    it will return product of two numbers
    a: number one
    b: number two

    """
    return a * b
```

```python
add_numbers.invoke({'a': 10, 'b': 2})
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers, mul_numbers],
    system_prompt = 'You are a math teacher and use tools for calculation'
)
```

```python
response = agent.invoke({'messages': HumanMessage(content='What is 2+3')})
```

```python
print(response['messages'][-1].content)
```

```python
response = agent.invoke({'messages': HumanMessage(content='What is 2+3*2')})
```

```python
print(response['messages'][-1].content)
```

```python
response = agent.invoke({'messages': HumanMessage(content='what is 2+5')})
```

```python
print(response['messages'][-1].content)
```

```python
from langgraph.checkpoint.memory import MemorySaver
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers, mul_numbers],
    system_prompt='You are a maths teacher you can use this tools for calculation.',
    checkpointer=MemorySaver()
)
```

```python
response = agent.invoke({'messages': HumanMessage(content='what is my name?')},
                        {'configurable': {'thread_id': 'id1'}})
```

```python
print(response['messages'][-1].content)
```

```python
state = agent.get_state({'configurable': {'thread_id': 'id1'}})
```

```python
for message in state.values['messages']:
    print(f'{message.type}: {message.content}')
    print()
```
